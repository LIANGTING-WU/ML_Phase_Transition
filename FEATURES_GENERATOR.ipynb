```python
import pandas as pd
from IPython.display import display
import numpy as np

data = [
  ['1', 'NO', 0.75, 0.1, {'Mn':2/3, 'Co': 1/6,'Ni': 1/6,'O': 2.0}],
  ['2', 'NO', 0.6, 0.1, {'Li':0.15, 'Co':0.15, 'Mn':0.7, 'O':2.0}],
  ['3', 'YES', 0.85, 0.05, {'Mn':3/4, 'Ni': 1/4, 'O': 2.0}],
  ['4', 'NO', 0.67, 0.11, {'Li':0.22, 'Mn':0.78, 'O':2.0}],
  ['5', 'YES', 0.75, 0.15, {'Li':0.15, 'Ni':0.15, 'Mn':0.70, 'O':2.0}],
  ['6', 'YES', 0.6, 0.09, {'Mg': 0.2, 'Ru': 0.2, 'Mn': 0.6, 'O': 2.0}],
  ['7', 'NO',  0.6, 0.09, {'Mg': 0.2, 'Mn': 0.8, 'O': 2.0}],
  ['8', 'NO',  0.67, 0.07, {'Li': 0.1, 'Co': 0.1, 'Cu': 0.1, 'Ni': 0.1, 'Ti': 0.1, 'Mn': 0.5, 'O': 2.0}],
  ['9', 'YES', 0.73, 0.16, {'Li': 0.06, 'Cu': 0.06, 'Ni': 0.21, 'Ti': 0.07, 'Mn': 0.6, 'O': 2.0}],
  ['10', 'NO', 0.67, 0.2, {'Li': 0.2, 'Ni': 0.2, 'Mn': 0.6, 'O': 2.0}],
  ['11', 'NO',  0.78, 0.15, {'Ni': 1/6, 'Co': 1/2, 'Mn': 1/3, 'O': 2.0}],
  ['12', 'NO',  0.6, 0.08, {'Li': 0.15, 'Fe': 0.15, 'Mn': 0.7, 'O': 2.0}],
  ['13', 'YES', 0.55, 0.05, {'Ni': 0.1, 'Fe': 0.1, 'Mn': 0.8, 'O': 2.0}],
  ['14', 'NO',  0.615, 0, {'Mg': 0.154, 'Ti': 0.154, 'Mn': 0.615, 'O': 2.0}],
  ['15', 'NO',  0.67, 0.26, {'Cu': 0.28, 'Mn': 0.72, 'O': 2.0}],
  ['16', 'YES', 0.67, 0.20, {'Ni': 0.15, 'Zn': 0.15, 'Mn': 0.7, 'O': 2.0}],
  ['17', 'NO',  0.66, 0.08, {'Ni': 0.22, 'Cu': 0.06, 'Zn': 0.05, 'Ti':0.01, 'Mn': 0.66, 'O': 2.0}],
  ['18', 'NO',  0.87, 0.47, {'Li': 0.1, 'Mg': 0.1, 'Cu':0.1, 'Mn': 0.7, 'O': 2.0}],
  ['19', 'NO',  0.75, 0.27, {'Ni': 0.17, 'Cu': 0.08, 'Mn': 0.75, 'O': 2.0}],
  ['20', 'NO',  0.75, 0.08, {'Li': 1/4, 'Mn': 3/4, 'O': 2.0}],
  ['21', 'YES', 0.67, 0, {'Mn': 1.0, 'O': 2.0}],
  ['22', 'YES', 0.67, 0.15, {'Ni': 0.2, 'Mn': 0.8, 'O': 2.0}],
  ['23', 'NO', 0.8, 0.21, {'Li': 0.18, 'Mg': 0.03, 'Cu': 0.15, 'Mn': 0.67, 'O': 2.0}],
  ['24', 'YES', 0.67, 0.09, {'Ni': 1/6, 'Fe': 1/3, 'Mn': 1/2, 'O': 2.0}],
  ['25', 'NO',  0.67, 0.15, {'Ti': 0.1, 'Fe': 0.198, 'Mn': 0.702, 'O': 2.0}],
  ['26', 'YES', 0.67, 0.2, {'Zn': 2/9, 'Mn': 7/9, 'O': 2.0}],
  ['27', 'YES',  0.6, 0.07, {'Mg': 0.2, 'Co': 0.2, 'Mn': 0.6, 'O': 2.0}],
  ['28', 'NO',  0.7, 0.09, {'Li': 0.1, 'Mg': 0.05, 'Ni': 0.15, 'Mn': 0.7, 'O': 2.0}],
  ['29', 'NO',  0.67, 0.29, {'Co': 0.2, 'Mn': 0.8, 'O': 2.0}],
  ['30', 'NO', 0.67, 0.24, {'Zn': 0.3, 'Mn': 0.7, 'O': 2.0}],
  ['31', 'YES', 0.67, 0.1, {'Mg': 0.28, 'Mn': 0.72, 'O': 2.0}],
  ['32', 'YES', 0.67, 0.13, {'Ni': 0.26, 'Zn': 0.07, 'Mn': 0.67, 'O': 2.0}],
  ['33', 'NO',  0.67, 0.2, {'Ni': 2/8, 'Li': 1/8, 'Mn': 5/8, 'O': 2.0}],
  ['34', 'NO',  0.57, 0.24, {'Mn': 6/7, 'O': 2.0}],
  ['35', 'YES', 0.67, 0.22, {'Ni': 2/8, 'Mn': 6/8, 'O': 2.0}],
  ['36', 'NO',  0.66, 0.21, {'Li': 0.18, 'Fe': 0.12, 'Mn': 0.7, 'O': 2.0}],
  ['37', 'YES', 0.67, 0.05, {'Ni': 0.35, 'Co': 0.1, 'Mn': 0.5, 'Al': 0.05, 'O': 2.0}],
  ['38', 'NO',  0.66, 0.09, {'Li': 0.24, 'Ca': 0.03, 'Mn': 0.76, 'O': 2.0}],
  ['39', 'YES', 0.67, 0.23, {'Mg': 0.05, 'Li': 0.05, 'Ni': 0.297,  'Mn': 0.603, 'O': 2.0}],
  ['40', 'YES', 0.67, 0.2, {'Mg': 0.1, 'Ni': 0.23, 'Mn': 0.65, 'Sn': 0.02, 'O': 2.0}],
  ['41', 'NO',  1.0, 0.43, {'Co': 0.1, 'Mn': 0.9, 'O': 2.0}],
  ['42', 'YES', 0.67, 0.19, {'Ni': 0.2, 'Mg': 0.1, 'Mn': 0.7, 'O': 2.0}],
  ['43', 'YES', 0.67, 0.05, {'Ni': 1/3, 'Mn': 2/3, 'O': 2.0}],
  ['44', 'YES', 0.67, 0.12, {'Fe': 1/2, 'Mn': 1/2, 'O': 2.0}],
  ['45', 'YES',  0.66, 0.01, {'Ni': 0.33, 'Ru': 0.3 ,'Mn': 0.36, 'O': 2.0}],
  ['46', 'NO',  0.67, 0.36, {'Co': 0.5, 'Mn': 0.5, 'O': 2.0}],
  ['47', 'YES', 0.66, 0.0, {'Li': 0.22, 'Ru': 0.78, 'O': 2.0}],
  ['48', 'NO', 0.67, 0.02, {'Cu': 0.15, 'Ti': 0.15, 'Mn': 0.7, 'O': 2.0}],
  ['49', 'NO',  0.67, 0.33, {'Ni': 1/6, 'Mn': 2/3, 'Cu': 1/9, 'Mg': 1/18, 'O': 2.0}],
  ['50', 'YES',  0.6, 0.0, {'Co': 0.2, 'Ti': 0.2, 'Ni': 0.2, 'Ru': 0.2, 'Mn': 0.2, 'O': 2.0}],
  ['51', 'YES', 0.75, 0.03, {'Ni': 0.22, 'Co': 0.78, 'O': 2.0}],
  ['52', 'YES',  0.67, 0.125, {'Mg': 1/4, 'Mn': 7/12, 'Co': 1/6, 'O': 2.0}],
  ['53', 'YES', 0.6, 0.17, {'Ru': 0.297, 'Ni': 0.303, 'Mn': 0.4, 'O': 2.0}],
  ['54', 'NO',  1.00, 0.33, {'Cr': 0.6, 'Ti': 0.4, 'O': 2.0}],
  ['55', 'NO',  0.70, 0.39, {'Co': 2/3, 'Mn': 2/9, 'Ni': 1/9, 'O': 2.0}],
  ['56', 'NO', 0.67, 0.22, {'Ni': 1/3, 'Mn': 1/3, 'Ti': 1/3, 'O': 2.0}],
  ['57', 'NO', 0.7, 0.42, {'Fe': 0.2, 'Mn': 0.8, 'O': 2.0}],
  ['58', 'YES', 0.7, 0.1, {'Fe': 0.4, 'Mn': 0.4, 'Co': 0.2, 'O': 2.0}],
  ['59', 'NO',  0.8, 0.37, {'Co': 0.8, 'Ti': 0.2, 'O': 2.0}],
  ['60', 'YES', 0.7, 0.22, {'Ni': 0.202, 'Ru': 0.298, 'Mn': 0.4, 'O': 2.0}],
  ['61', 'NO',  0.67, 0.15, {'Mn': 5/6, 'O': 2.0}],
  ['62', 'YES', 0.62, 0.05, {'K': 0.05, 'Mg': 2/9, 'Cu': 1/9, 'Mn': 2/3, 'O': 2.0}],
  ['63', 'YES', 0.67, 0.0, {'Ag': 0.06, 'Mn': 0.94, 'O': 2.0}],
  ['64', 'NO',  0.67, 0.3, {'Co': 0.5, 'Ti': 0.5, 'O': 2.0}],
  ['65', 'YES', 0.67, 0.0, {'Ni': 0.21, 'Co': 0.24, 'Mn':0.55, 'O': 2.0}],
  ['66', 'YES', 0.67, 0.05, {'Ni': 0.18, 'Co': 0.24, 'Mg': 0.03, 'Ti': 0.1, 'Mn': 0.45, 'O': 2.0}],
  ['67', 'NO',  0.67, 0.05, {'Ni': 0.18, 'Co': 0.18, 'Mg': 0.03, 'Ti': 0.1, 'Al': 0.04, 'Fe': 0.02, 'Mn': 0.45, 'O': 2.0}],
  ['68', 'YES',  0.67, 0.1, {'Ni': 1/4, 'Cu': 1/12, 'Mn': 2/3, 'O': 2.0}],
  ['69', 'NO',  0.67, 0.24, {'Ni': 0.19, 'Cu': 0.07, 'Zn': 0.07, 'Ti': 0.07, 'Mn': 0.60, 'O': 2.0}],
  ['70', 'YES',  0.67, 0.14, {'Fe': 1/4, 'Co': 1/4, 'Mn': 1/2, 'O': 2.0}],
  ['71', 'YES',  0.75, 0.3, {'Fe': 0.20, 'Cu': 0.20, 'Mn': 0.60, 'O': 2.0}],
  ['72', 'NO',  0.85, 0.4, {'Li': 0.17, 'Ni': 0.21, 'Mn': 0.64, 'O': 2.0}],
  ['73', 'NO',  0.6, 0.05, {'Li': 0.11, 'Fe': 0.27, 'Mn': 0.62, 'O': 2.0}],
  ['74', 'NO',  0.78, 0.1, {'Ni': 0.23, 'Mn': 0.69, 'O': 2.0}],
  ['75', 'YES',  0.67, 0.0, {'Mg': 0.1, 'Zn': 0.1, 'Mn': 0.8, 'O': 2.0}],
  ['76', 'YES',  0.67, 0.0, {'Mg': 1/3, 'Ti': 1/6, 'Mn': 1/2, 'O': 2.0}],
  ['77', 'YES',  0.67, 0.13, {'Ni': 1/4, 'Mn': 1/2, 'Ti': 1/6, 'Zn': 1/12, 'O': 2.0}],
  ['78', 'YES',  0.67, 0.18, {'Ni': 1/3, 'Mn': 1/2, 'Ti': 1/6, 'O': 2.0}],
  ['79', 'YES',  0.67, 0.0, {'Fe': 1/3, 'Mn': 2/3, 'O': 2.0}],
  ['80', 'NO',  0.67, 0.06, {'Fe': 1/3, 'Mn': 0.57, 'Ti': 0.1, 'O': 2.0}],
  ['81', 'NO',  0.85, 0.35, {'Li': 0.12, 'Ni': 0.22, 'Mn': 0.66, 'O': 2.0}],
  ['82', 'NO',  0.78, 0.40, {'Ni': 0.31, 'Nb': 0.02, 'Mn': 0.67, 'O': 2.0}],
  ['83', 'NO',  0.67, 0.0, {'Mg': 0.1, 'Fe': 0.1, 'Mn': 0.8, 'O': 2.0}],
  ['84', 'YES',  0.6, 0.0, {'Al': 0.11, 'Ni': 0.22, 'Mn': 0.66, 'O': 2.0}],
  ['85', 'YES',  0.6, 0.0, {'Al': 0.11, 'Ni': 0.11, 'Fe': 0.11, 'Mn': 0.66, 'O': 2.0}],
  ['86', 'NO',  0.6, 0.2, {'Al': 0.11, 'Fe': 0.22, 'Mn': 0.66, 'O': 2.0}],
  ['87', 'NO',  1.0, 0.47, {'Ni': 2/9, 'Mg': 1/9, 'Co': 1/3, 'Mn': 1/3, 'O': 2.0}],
  ['88', 'YES', 0.67, 0.35, {'Co':1/3, 'Ti':2/3, 'O':2.0}],
  ['89', 'NO', 0.62, 0.12, {'Mn':0.67, 'Ni': 0.23, 'Cu':0.05, 'Mg':0.07, 'Ti':0.01, 'O': 2.0}],
  ['90', 'NO', 0.67, 0.16, {'Li':2/9, 'Mn':53/72, 'Ti':1/24, 'O':2.0}],
  ['91', 'YES', 0.67, 0.22, {'Li':2/9,'Mn':7/9, 'O':2.0}],
  ['92', 'YES', 0.67, 0.19, {'Li':1/6, 'Co':1/6, 'Mn':2/3, 'O':2.0}],
  ['93', 'YES', 0.67, 0.21, {'Co':1/3 , 'Mn':2/3, 'O':2.0}],
  ['94', 'YES', 0.72, 0.26, {'Mn':0.57, 'Ti': 0.10, 'Li':0.05, 'Ni':0.28, 'O': 2.0}],
  ['95', 'YES', 0.67, 0.15, {'Fe':0.25 , 'Mg':0.15, 'Mn':0.6, 'O':2.0}],
  ['96', 'NO', 7/9, 0.4, {'Cu':2/9 , 'Fe':1/9, 'Mn':2/3, 'O':2.0}],
  ['97', 'NO', 0.7, 0.53, {'Li':0.06 , 'Zn':0.06, 'Ni':0.21, 'Mn':0.67, 'O':2.0}],
  ['98', 'NO', 0.65, 0.33, {'Li':0.08 , 'Cu':0.08, 'Fe':0.24, 'Mn':0.6, 'O':2.0}],
  ['99', 'YES', 0.65, 0.256, {'Fe':0.4 , 'Mn':0.6, 'O':2.0}],
  ['100', 'NO', 0.75, 0, {'Li':0.08 , 'Cu':0.08, 'Fe':0.24, 'Mn':0.6, 'O':2.0}],
  ['101', 'NO', 0.76, 0.24, {'Cu':0.22 ,'Fe':0.30, 'Mn':0.48, 'O':2.0}]
]

df = pd.DataFrame(data, columns=['sample_id','phase_change','initial_na','final_na','transition_layer'])

# Radii dictionary
radii = {
    "Li": {1: 76},
    "Na": {1: 102},
    "Nt": {1: 102},
    "K": {1: 138},
    "Ca": {2: 100},
    "O":  {-2: 140},
    "Mg": {2: 72},
    "Ti": {2: 86, 3: 67, 4: 60.5},
    "V":  {2: 79, 3: 64, 4: 58, 5: 54},
    "Cr": {2: 73, 3: 61.5, 4: 55, 5: 49, 6: 44},
    "Mn": {2: 67, 3: 58, 4: 53},
    "Fe": {2: 61, 3: 55},
    "Co": {2: 65, 3: 54.5, 4: 53},
    "Ni": {2: 69, 3: 56, 4: 48},
    "Cu": {1: 77, 2: 73, 3: 54},
    "Zn": {2: 74},
    "Sr": {2: 118},
    "Ru": {3: 68, 4: 62},
    "Nb": {3: 72, 4: 68, 5: 64},
    "Te": {4: 97, 6: 56},
    "Mo": {3: 69, 4: 65, 5: 61, 6: 59},
    "Rh": {3: 66.5, 4: 60, 5: 55},
    "Lu": {3: 86.1},
    "Zr": {4: 72},
    "Pd": {1: 59, 2: 86, 3: 76, 4: 61.5},
    "Ir": {3: 68, 4: 62.5, 5: 57},
    "Pt": {2: 80, 4: 62.5, 5: 57},
    "Sn": {2: 109, 4: 69},
    "Bi": {3: 103, 5: 76},
    "Sb": {3: 76, 5: 60},
    "Al": {3: 53.5},
    "Er": {3: 103},
    "Tl": {3: 102.5},
    "Sc": {3: 88.5},
    "In": {3: 94},
    "Y":  {3: 104},
    "Ho": {3: 104.1},
    "Yb": {2: 116, 3: 100.8},
    "Pu": {3: 100, 4: 86},
    "Ag": {1: 115}
}


ionization_energies = {
    "Li": {1: 75.640097},
    "Na": {1: 47.28636},
    "Nt": {1: 47.28636},
    "K": {1: 31.62500},
    "Ca": {2: 50.91316},
    "O":  {-2: 140, -1: 0},
    "Mg": {2: 80.1436},
    "Ti": {2: 27.49171, 3: 43.26717, 4: 99.299},
    "V":  {2: 29.3111, 3: 46.709, 4: 65.28165, 5: 128.125},
    "Cr": {2: 30.959, 3: 49.16, 4: 69.46, 5: 90.6349, 6: 160.29},
    "Mn": {2: 33.668, 3: 51.21, 4: 72.41},
    "Fe": {2: 30.651, 3: 54.91},
    "Co": {2: 33.5, 3: 51.27, 4: 79.5},
    "Ni": {2: 35.187, 3: 54.92, 4: 76.06},
    "Cu": {1: 20.29239, 2: 36.841, 3: 57.38},
    "Zn": {2: 39.7233},
    "Sr": {2: 42.88353},
    "Ru": {3: 45, 4: 59},
    "Nb": {3: 37.611, 4: 50.5728, 5: 102.069},
    "Te": {4: 59.3, 6: 124.2},
    "Mo": {3: 40.33, 4: 54.417, 5: 68.82704, 6: 125.638},
    "Rh": {3: 42, 4: 63, 5: 80},
    "Lu": {3: 45.249},
    "Zr": {4: 80.348},
    "Pd": {1: 19.43, 2: 32.93, 3: 46, 4: 61},
    "Ir": {3: 40, 4: 57, 5: 72},
    "Pt": {2: 29, 4: 56, 5: 75},
    "Sn": {2: 30.506, 4: 77.03},
    "Bi": {3: 103, 5: 76},
    "Sb": {3: 43.804, 5: 99.51},
    "Al": {3: 119.9924},
    "Er": {3: 42.42},
    "Tl": {3: 51.14},
    "Sc": {3: 73.4894},
    "In": {3: 55.45},
    "Y":  {3: 60.6072},
    "Ho": {3: 42.52},
    "Yb": {2: 25.053, 3: 43.61},
    "Pu": {3: 35, 4: 49},
    "Ag": {1: 21.4844}
}

def initial_state(el):
    oks = sorted(ionization_energies.get(el, {}).keys())
    if 3 in oks: return 3
    if 2 in oks: return 2
    if 4 in oks: return 4
    return oks[0] if oks else 0

def get_adjacent_state(el, st, ion_dict, direction='up'):
    oks = sorted(ion_dict.get(el, {}).keys())
    if not oks: return None
    try:
        idx = oks.index(st)
    except ValueError: return None
    if direction == 'up' and idx < len(oks) - 1: return oks[idx + 1]
    if direction == 'down' and idx > 0: return oks[idx - 1]
    return None

def compute_charge(state_sc):
    return sum(ox * conc for (el, ox), conc in state_sc.items())

def adjust_charge_to_target(sc_input, target_charge):
    sc = sc_input.copy()
    tol = 1e-6

    while True:
        current_charge = compute_charge(sc)
        diff = target_charge - current_charge

        if abs(diff) <= tol:
            break

        if diff > tol: # Oxidation
            cands = []
            for (el, st), avail in sc.items():
                if avail <= tol or el == 'O': continue
                next_st = get_adjacent_state(el, st, ionization_energies, 'up')
                if next_st and next_st > st:
                    ie_next = ionization_energies[el][next_st]
                    cands.append(((el, st, next_st), ie_next, avail, next_st - st))

            if cands:
                cands.sort(key=lambda x: x[1])
                best = cands[0]
                needed = diff
                conc_shift = needed / best[3]
                delta = min(best[2], conc_shift)

                sc[(best[0][0], best[0][1])] -= delta
                sc[(best[0][0], best[0][2])] = sc.get((best[0][0], best[0][2]), 0) + delta

            else:
                o2_conc = sc.get(('O', -2), 0)
                if o2_conc > tol:
                    needed = diff
                    delta = min(o2_conc, needed)
                    sc[('O', -2)] -= delta
                    sc[('O', -1)] = sc.get(('O', -1), 0) + delta
                else:
                    break

        elif diff < -tol: # Reduction
            diff = abs(diff)
            cands = []
            for (el, st), avail in sc.items():
                if avail <= tol or el == 'O': continue
                prev_st = get_adjacent_state(el, st, ionization_energies, 'down')
                if prev_st is not None and prev_st < st:
                    ie_curr = ionization_energies[el][st]
                    cands.append(((el, st, prev_st), ie_curr, avail, st - prev_st))

            if not cands:
                break
            cands.sort(key=lambda x: x[1], reverse=True)
            best = cands[0]

            needed = diff
            conc_shift = needed / best[3]
            delta = min(best[2], conc_shift)

            sc[(best[0][0], best[0][1])] -= delta
            sc[(best[0][0], best[0][2])] = sc.get((best[0][0], best[0][2]), 0) + delta

    return sc

def compute_TM_metrics_safe(state_sc, radii, ionization_energies):
    tm_items = [(el, st, conc)
                for (el, st), conc in state_sc.items()
                if el not in ['Na', 'O']]
    valid = []
    for el, st, conc in tm_items:
        r = radii.get(el, {}).get(st)
        ie = ionization_energies.get(el, {}).get(st)
        if r is None: continue
        if ie is None: ie = 0.0
        valid.append((el, st, conc, r, ie))

    n_TM = sum(c for el, st, c, r, ie in valid)
    if n_TM == 0:
        return dict.fromkeys(['n_TM','r_TM','q_TM','p_TM','I_TM','S_TM'], 0)

    r_TM = sum(c * r   for el, st, c, r, ie in valid) / n_TM
    q_TM = sum(c * st  for el, st, c, r, ie in valid) # Total Charge provided by TM
    p_TM = sum(c * (st/r) for el, st, c, r, ie in valid)
    I_TM = sum(c * ie  for el, st, c, r, ie in valid)
    S_TM = -sum(c * np.log(c) for el, st, c, r, ie in valid if c > 1e-9)

    return {
        'n_TM': n_TM, 'r_TM': r_TM, 'q_TM': q_TM,
        'p_TM': p_TM, 'I_TM': I_TM, 'S_TM': S_TM
    }

def compute_valence_counts(sc):
    monovalent = sum(conc for (el, st), conc in sc.items()
                     if el not in ['Na', 'O'] and set(ionization_energies.get(el, {}).keys()) == {1})
    divalent   = sum(conc for (el, st), conc in sc.items()
                     if el not in ['Na', 'O'] and set(ionization_energies.get(el, {}).keys()) == {2})
    trivalent  = sum(conc for (el, st), conc in sc.items()
                     if el not in ['Na', 'O'] and set(ionization_energies.get(el, {}).keys()) == {3})
    return {'monovalent': monovalent, 'divalent': divalent, 'trivalent': trivalent}

# ==========================================
# 4. Main Execution
# ==========================================

initial_states, final_states = [], []

for _, row in df.iterrows():
    sc0 = {(el, initial_state(el)): conc for el, conc in row['transition_layer'].items()}
    sc0[('Na', 1)] = row['initial_na']
    sc_init = adjust_charge_to_target(sc0, target_charge=0)
    initial_states.append(sc_init)

    sc1 = sc_init.copy()
    sc1[('Na', 1)] = row['final_na']
    sc_fin = adjust_charge_to_target(sc1, target_charge=0)
    final_states.append(sc_fin)

df['initial_state_sc'] = initial_states
df['final_state_sc'] = final_states

df_charges = pd.DataFrame({
    'sample_id':      df['sample_id'],
    'initial_charge': df['initial_state_sc'].apply(compute_charge),
    'final_charge':   df['final_state_sc'].apply(compute_charge),
})

init_df = pd.DataFrame([ {f"{el}_{st}":conc for (el,st),conc in sc.items()} for sc in df['initial_state_sc'] ])
fin_df  = pd.DataFrame([ {f"{el}_{st}":conc for (el,st),conc in sc.items()} for sc in df['final_state_sc'] ])

init_metrics = df['initial_state_sc'].apply(lambda sc: compute_TM_metrics_safe(sc, radii, ionization_energies)).apply(pd.Series).add_prefix('init_')
fin_metrics = df['final_state_sc'].apply(lambda sc: compute_TM_metrics_safe(sc, radii, ionization_energies)).apply(pd.Series).add_prefix('final_')

init_valence = pd.concat([df['sample_id'], df['initial_state_sc'].apply(compute_valence_counts).apply(pd.Series).add_prefix('init_')], axis=1)

combined_base = df_charges.merge(pd.concat([df[['sample_id']], init_metrics, fin_metrics], axis=1), on='sample_id') \
                          .merge(init_valence, on='sample_id')

init_expanded = pd.concat([df[['sample_id', 'initial_na', 'final_na', 'phase_change']], init_df.add_prefix('init_state_')], axis=1)
fin_expanded = pd.concat([df[['sample_id']], fin_df.add_prefix('final_state_')], axis=1)

combined_full = (combined_base.merge(init_expanded, on='sample_id').merge(fin_expanded, on='sample_id'))

combined_full = combined_full.fillna(0)

display(combined_full)
```

```python
def keep_column(col_name):
    base_cols = ['sample_id', 'phase_change', 'initial_na', 'final_na']
    if col_name in base_cols:
        return True

    if ('init_' in col_name or 'final_' in col_name) and '_state_' not in col_name:
        return True

    if '_state_' in col_name:
        if 'Na' in col_name or 'O' in col_name:
            return True

    return False

simple_columns = [c for c in combined_full.columns if keep_column(c)]
df_simple = combined_full[simple_columns].copy()

df_simple['phase_change'] = df_simple['phase_change'].map({'YES': 1, 'NO': 0})

cols = df_simple.columns.tolist()
sorted_cols = [c for c in cols if 'final' not in c.lower()] + [c for c in cols if 'final' in c.lower()]
df_simple = df_simple[sorted_cols]

display(df_simple.head())

df_simple.to_csv('Phase-Transition.csv', index=False)
```
